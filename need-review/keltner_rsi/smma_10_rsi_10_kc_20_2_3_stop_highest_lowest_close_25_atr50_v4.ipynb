# 3 BACKTESTER
# Strategy:
#   Long signal:
#     Close(t-1) > SMMA10_Close(t-1)
#     RSI(t-2) < RSI(t-1)
#     Open(t) > KC_Upper(20,dev=2.3, shift=1)
#   Short signal:
#     Close(t-1) < SMMA10_Close(t-1)
#     RSI(t-2) > RSI(t-1)
#     Open(t) < KC_Lower(20,dev=2.3, shift=1)
#
# Entry STOP:
#   BuyStop  = Highest(Close,25,shift=3) + 2.4 * ATR50(shift=3)
#   SellStop = Lowest (Close,25,shift=3) - 2.4 * ATR50(shift=3)
#   Pending expiration = 4 bars
#
# Exit:
#   SL_dist = 2.1 * ATR20(shift=1)
#   TP_dist = 4.5 * ATR30(shift=1)
#   ExitAfterBars = 14
# =========================================================

```python
import requests
import pandas as pd
import numpy as np
from datetime import datetime
import time
import matplotlib.pyplot as plt
import matplotlib.dates as mdates


# =========================================================
# 1) FETCH BINANCE DATA
# =========================================================
def str_to_millis(date_str: str) -> int:
    dt = datetime.strptime(date_str, "%d/%m/%Y")
    return int(dt.timestamp() * 1000)

def fetch_binance_kline_df(symbol: str, interval: str, start_str: str, end_str: str) -> pd.DataFrame:
    base_url = "https://api.binance.com/"
    api_url = "api/v3/klines"
    url = base_url + api_url

    start_time = str_to_millis(start_str)
    end_dt = datetime.strptime(end_str, "%d/%m/%Y").replace(hour=23, minute=59, second=59)
    end_time = int(end_dt.timestamp() * 1000)

    all_klines = []
    print(f"⏳ Downloading {symbol} {interval}...")

    while start_time < end_time:
        params = {
            "symbol": symbol.upper(),
            "interval": interval,
            "startTime": start_time,
            "endTime": end_time,
            "limit": 1000
        }
        try:
            r = requests.get(url, params=params, timeout=10)
            if r.status_code != 200:
                print("Error:", r.status_code, r.text)
                break

            data = r.json()
            if not data:
                break

            all_klines.extend(data)
            start_time = data[-1][0] + 1
            time.sleep(0.05)

        except Exception as e:
            print("Fetch error:", e)
            break

    if not all_klines:
        return pd.DataFrame()

    df = pd.DataFrame(all_klines, columns=[
        "timestamp", "Open", "High", "Low", "Close", "Volume",
        "close_time", "quote_asset_volume", "num_trades",
        "taker_buy_base_volume", "taker_buy_quote_volume", "ignore"
    ])
    df = df[["timestamp", "Open", "High", "Low", "Close", "Volume"]]
    df["timestamp"] = pd.to_datetime(df["timestamp"], unit="ms")
    df.set_index("timestamp", inplace=True)
    df = df.astype(float)
    return df


# =========================================================
# 2) INDICATORS
#    - SMMA(10) on CLOSE (RMA / Wilder smoothing)
#    - RSI(10) on CLOSE
#    - ATR(20), ATR(30), ATR(50)
#    - Keltner Channel(20, dev=2.3)
#    - Highest/Lowest Close window=25
# =========================================================
def calculate_indicators(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()

    # --- True Range ---
    prev_close = data["Close"].shift(1)
    tr = pd.concat([
        (data["High"] - data["Low"]),
        (data["High"] - prev_close).abs(),
        (data["Low"] - prev_close).abs(),
    ], axis=1).max(axis=1)
    data["TR"] = tr

    # --- Wilder ATR (EMA alpha=1/n) ---
    def calc_atr(tr_series: pd.Series, period: int) -> pd.Series:
        return tr_series.ewm(alpha=1/period, min_periods=period, adjust=False).mean()

    data["ATR_20"] = calc_atr(data["TR"], 20)
    data["ATR_30"] = calc_atr(data["TR"], 30)
    data["ATR_50"] = calc_atr(data["TR"], 50)

    # --- SMMA(10) on CLOSE (RMA) ---
    # SMMA ~= Wilder moving average = ewm(alpha=1/n, adjust=False)
    data["SMMA_10_Close"] = data["Close"].ewm(alpha=1/10, min_periods=10, adjust=False).mean()

    # --- RSI(10) on CLOSE (rolling mean style like your template) ---
    def calc_rsi(series: pd.Series, period: int) -> pd.Series:
        delta = series.diff()
        gain = delta.where(delta > 0, 0.0).rolling(window=period).mean()
        loss = (-delta.where(delta < 0, 0.0)).rolling(window=period).mean()
        rs = gain / loss
        return 100 - (100 / (1 + rs))

    data["RSI_10_Close"] = calc_rsi(data["Close"], 10)

    # --- Keltner Channel(20, dev=2.3) ---
    ema_20 = data["Close"].ewm(span=20, adjust=False).mean()
    atr_20_kc = calc_atr(data["TR"], 20)
    data["KC_Upper_20_2p3"] = ema_20 + 2.3 * atr_20_kc
    data["KC_Lower_20_2p3"] = ema_20 - 2.3 * atr_20_kc

    # --- Highest/Lowest Close window=25 ---
    data["Highest_Close_25"] = data["Close"].rolling(window=25).max()
    data["Lowest_Close_25"] = data["Close"].rolling(window=25).min()

    data.dropna(inplace=True)
    return data


# =========================================================
# 3) BACKTESTER
# Strategy:
#   Long signal:
#     Close(t-1) > SMMA10_Close(t-1)
#     RSI(t-2) < RSI(t-1)
#     Open(t) > KC_Upper(20,dev=2.3, shift=1)
#   Short signal:
#     Close(t-1) < SMMA10_Close(t-1)
#     RSI(t-2) > RSI(t-1)
#     Open(t) < KC_Lower(20,dev=2.3, shift=1)
#
# Entry STOP:
#   BuyStop  = Highest(Close,25,shift=3) + 2.4 * ATR50(shift=3)
#   SellStop = Lowest (Close,25,shift=3) - 2.4 * ATR50(shift=3)
#   Pending expiration = 4 bars
#
# Exit:
#   SL_dist = 2.1 * ATR20(shift=1)
#   TP_dist = 4.5 * ATR30(shift=1)
#   ExitAfterBars = 14
# =========================================================
class SMMA_RSI_KC_Breakout_Backtester:
    def __init__(self, df: pd.DataFrame, initial_balance: float = 10000.0):
        self.df = df
        self.initial_balance = float(initial_balance)
        self.balance = float(initial_balance)

        self.equity_curve = []
        self.trades = []

        self.position = None  # "LONG" / "SHORT" / None
        self.entry_price = 0.0
        self.sl = 0.0
        self.tp = 0.0
        self.qty = 0.0
        self.entry_bar_idx = 0

        self.pending_order = None

        # Params
        self.price_entry_mult = 2.4
        self.exit_after_bars = 14
        self.pending_expiration = 4

        self.sl_coef = 2.1
        self.tp_coef = 4.5

        # MM
        self.risk_pct = 0.01 
        # self.mm_max_lots = 100.0
        # self.mm_decimals = 2

        # self.mm_max_notional = 2000.0   # ví dụ: tối đa 2000 USDT mỗi lệnh
        # self.mm_step_size = 0.001       # bạn nên set theo symbol (hoặc auto)
        # self.mm_min_qty = 0.0
    def run(self):
        idxs = self.df.index

        for i in range(len(self.df)):
            bar = self.df.iloc[i]
            t = idxs[i]

            # -----------------------------
            # A) Manage open position
            # -----------------------------
            if self.position is not None:
                held = i - self.entry_bar_idx

                # Time exit
                if held >= self.exit_after_bars:
                    self.close_position(bar["Close"], t, "Time Exit")
                    self.append_equity(t)
                    continue

                if self.position == "LONG":
                    if bar["Low"] <= self.sl:
                        self.close_position(self.sl, t, "Stop Loss")
                        self.append_equity(t)
                        continue
                    if bar["High"] >= self.tp:
                        self.close_position(self.tp, t, "Take Profit")
                        self.append_equity(t)
                        continue

                elif self.position == "SHORT":
                    if bar["High"] >= self.sl:
                        self.close_position(self.sl, t, "Stop Loss")
                        self.append_equity(t)
                        continue
                    if bar["Low"] <= self.tp:
                        self.close_position(self.tp, t, "Take Profit")
                        self.append_equity(t)
                        continue

            # -----------------------------
            # B) Manage pending order
            # -----------------------------
            if self.position is None and self.pending_order is not None:
                if i >= self.pending_order["expire_idx"]:
                    self.pending_order = None
                else:
                    p = self.pending_order["price"]

                    if self.pending_order["type"] == "BUY_STOP":
                        if bar["High"] >= p:
                            fill = max(p, bar["Open"])
                            self.open_position("LONG", fill, i,
                                               self.pending_order["sl_dist"],
                                               self.pending_order["tp_dist"])
                            self.pending_order = None

                    elif self.pending_order["type"] == "SELL_STOP":
                        if bar["Low"] <= p:
                            fill = min(p, bar["Open"])
                            self.open_position("SHORT", fill, i,
                                               self.pending_order["sl_dist"],
                                               self.pending_order["tp_dist"])
                            self.pending_order = None

            # -----------------------------
            # C) Compute signals -> place pending
            # Need:
            #   - shift=1 for MA/KC/ATR20/ATR30
            #   - shift=2 vs shift=1 for RSI slope
            #   - shift=3 for Highest/Lowest and ATR50 entry
            # => minimal i >= 3
            # -----------------------------
            if self.position is None and self.pending_order is None:
                if i < 3:
                    self.append_equity(t)
                    continue

                open_0 = bar["Open"]

                # Trend filter uses shift=1:
                close_1 = self.df.iloc[i-1]["Close"]
                smma10_1 = self.df.iloc[i-1]["SMMA_10_Close"]

                # RSI slope: shift=2 vs shift=1
                rsi_2 = self.df.iloc[i-2]["RSI_10_Close"]
                rsi_1 = self.df.iloc[i-1]["RSI_10_Close"]

                # KC shift=1:
                kc_up_1 = self.df.iloc[i-1]["KC_Upper_20_2p3"]
                kc_lo_1 = self.df.iloc[i-1]["KC_Lower_20_2p3"]

                long_signal = (close_1 > smma10_1) and (rsi_2 < rsi_1) and (open_0 > kc_up_1)
                short_signal = (close_1 < smma10_1) and (rsi_2 > rsi_1) and (open_0 < kc_lo_1)

                if (not long_signal) and (not short_signal):
                    self.append_equity(t)
                    continue

                # Entry uses Highest/Lowest(Close,25, shift=3) => take at i-3
                highest_close_3 = self.df.iloc[i-3]["Highest_Close_25"]
                lowest_close_3 = self.df.iloc[i-3]["Lowest_Close_25"]

                # Entry buffer uses ATR50 shift=3
                atr50_3 = self.df.iloc[i-3]["ATR_50"]

                # SL/TP distance uses shift=1
                atr20_1 = self.df.iloc[i-1]["ATR_20"]
                atr30_1 = self.df.iloc[i-1]["ATR_30"]
                sl_dist = self.sl_coef * atr20_1
                tp_dist = self.tp_coef * atr30_1

                if long_signal:
                    entry = highest_close_3 + self.price_entry_mult * atr50_3
                    self.pending_order = {
                        "type": "BUY_STOP",
                        "price": float(entry),
                        "sl_dist": float(sl_dist),
                        "tp_dist": float(tp_dist),
                        "expire_idx": i + self.pending_expiration
                    }
                elif short_signal and (not long_signal):
                    entry = lowest_close_3 - self.price_entry_mult * atr50_3
                    self.pending_order = {
                        "type": "SELL_STOP",
                        "price": float(entry),
                        "sl_dist": float(sl_dist),
                        "tp_dist": float(tp_dist),
                        "expire_idx": i + self.pending_expiration
                    }

            # -----------------------------
            # D) Append equity
            # -----------------------------
            self.append_equity(t)

    def append_equity(self, t):
        mtm = self.balance
        close_px = self.df.loc[t, "Close"]

        if self.position == "LONG":
            mtm += (close_px - self.entry_price) * self.qty
        elif self.position == "SHORT":
            mtm += (self.entry_price - close_px) * self.qty

        self.equity_curve.append({"time": t, "equity": mtm})

    def open_position(self, pos_type: str, price: float, bar_idx: int, sl_dist: float, tp_dist: float):
        self.position = pos_type
        self.entry_price = float(price)
        self.entry_bar_idx = int(bar_idx)

        if pos_type == "LONG":
            self.sl = self.entry_price - sl_dist
            self.tp = self.entry_price + tp_dist
        else:
            self.sl = self.entry_price + sl_dist
            self.tp = self.entry_price - tp_dist

        # Risk Fixed % Balance sizing
        risk_amount = self.balance * self.risk_pct
        sl_distance = abs(self.entry_price - self.sl)
        qty = (risk_amount / sl_distance) if sl_distance > 0 else 0.0

        # # --- CAP THEO NOTIONAL (USDT) ---
        # max_qty_by_notional = self.mm_max_notional / self.entry_price
        # qty = min(qty, max_qty_by_notional)

        # # --- SNAP THEO STEP SIZE ---
        # step = self.mm_step_size
        # qty = np.floor(qty / step) * step

        # # --- MIN QTY ---
        # if qty < self.mm_min_qty:
        #     qty = 0.0

        self.qty = float(qty)

    def close_position(self, price: float, t, reason: str):
        price = float(price)

        if self.position == "LONG":
            pnl = (price - self.entry_price) * self.qty
        else:
            pnl = (self.entry_price - price) * self.qty

        self.balance += pnl

        self.trades.append({
            "Exit Time": t,
            "Type": self.position,
            "Entry": self.entry_price,
            "Exit": price,
            "PnL": pnl,
            "Reason": reason,
            "Balance": self.balance
        })

        self.position = None
        self.qty = 0.0


# =========================================================
# 4) METRICS + PLOT
# =========================================================
def calculate_metrics(trades_df: pd.DataFrame, equity_data, initial_balance: float):
    eq_df = pd.DataFrame(equity_data)
    eq_df.set_index("time", inplace=True)
    eq_df = eq_df[~eq_df.index.duplicated(keep="last")]

    final_equity = eq_df["equity"].iloc[-1]
    total_return = (final_equity - initial_balance) / initial_balance

    days = (eq_df.index[-1] - eq_df.index[0]).days
    years = days / 365.25 if days > 0 else 0.0
    cagr = ((final_equity / initial_balance) ** (1/years) - 1) if years > 0 else 0.0

    running_max = eq_df["equity"].cummax()
    dd = (eq_df["equity"] - running_max) / running_max
    max_dd = dd.min()

    eq_df["returns"] = eq_df["equity"].pct_change()
    mean_ret = eq_df["returns"].mean()
    std_ret = eq_df["returns"].std()

    bars_per_year = (len(eq_df) / years) if years > 0 else 8760
    sharpe = (mean_ret / std_ret) * np.sqrt(bars_per_year) if std_ret != 0 else 0.0

    total_trades = len(trades_df)
    win_trades = len(trades_df[trades_df["PnL"] > 0])
    win_rate = win_trades / total_trades if total_trades > 0 else 0.0
    avg_pnl = trades_df["PnL"].mean() if total_trades > 0 else 0.0

    gross_profit = trades_df[trades_df["PnL"] > 0]["PnL"].sum()
    gross_loss = abs(trades_df[trades_df["PnL"] < 0]["PnL"].sum())
    profit_factor = (gross_profit / gross_loss) if gross_loss > 0 else 0.0

    return {
        "Final Equity": final_equity,
        "Total Return": total_return,
        "CAGR (approx)": cagr,
        "Max Drawdown": max_dd,
        "Sharpe": sharpe,
        "Trades": total_trades,
        "Winrate": win_rate,
        "Avg PnL/Trade": avg_pnl,
        "Profit Factor": profit_factor,
        "equity_df": eq_df
    }

def print_report_and_plot(metrics, symbol: str):
    print("\n" + "="*60)
    print("🎯 STRATEGY - SMMA(10) + RSI(10) + KC(20,2.3) + STOP(Highest/Lowest Close 25) + ATR50 buffer")
    print("="*60)
    print(f"     Final Equity: {metrics['Final Equity']:.4f}")
    print(f"     Total Return: {metrics['Total Return']:.4f}")
    print(f"    CAGR (approx): {metrics['CAGR (approx)']:.4f}")
    print(f"     Max Drawdown: {metrics['Max Drawdown']:.4f}")
    print(f"           Sharpe: {metrics['Sharpe']:.4f}")
    print(f"           Trades: {metrics['Trades']}")
    print(f"          Winrate: {metrics['Winrate']:.4f}")
    print(f"    Avg PnL/Trade: {metrics['Avg PnL/Trade']:.4f}")
    print(f"    Profit Factor: {metrics['Profit Factor']:.4f}")
    print("="*60)

    rr = (metrics["Profit Factor"] * (1 - metrics["Winrate"])) / metrics["Winrate"] if metrics["Winrate"] > 0 else 0.0
    print(f"\n💡 Estimated R:R: 1:{rr:.2f}")

    eq_df = metrics["equity_df"]
    plt.figure(figsize=(12, 6))
    plt.plot(eq_df.index, eq_df["equity"], linewidth=1.5)
    plt.fill_between(eq_df.index, eq_df["equity"], alpha=0.3)
    plt.title(f"{symbol} SMMA+RSI+KC Breakout (H1)", fontsize=14, fontweight="bold")
    plt.xlabel("Date")
    plt.ylabel("Equity ($)")
    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.tight_layout()
    plt.show()


# =========================================================
# 5) MAIN
# =========================================================
# if __name__ == "__main__":
#     symbol = "ETHUSDT"
#     interval = "1h"
#     start_date = "01/01/2025"
#     end_date = "5/02/2026"

#     df_raw = fetch_binance_kline_df(symbol, interval, start_date, end_date)

#     if len(df_raw) > 400:
#         print("✅ Data ready...")
#         df_calc = calculate_indicators(df_raw)

#         print("🚀 Running strategy...")
#         bt = SMMA_RSI_KC_Breakout_Backtester(df_calc, initial_balance=10000.0)
#         bt.run()

#         trades_df = pd.DataFrame(bt.trades)

#         if not trades_df.empty:
#             metrics = calculate_metrics(trades_df, bt.equity_curve, bt.initial_balance)
#             print_report_and_plot(metrics, symbol)

#             print("\n📊 Sample Trades:")
#             print(trades_df.head(10).to_string())

#             print("\n📈 Exit Reasons:")
#             print(trades_df["Reason"].value_counts())
#         else:
#             print("⚠️ No trades.")
#     else:
#         print("❌ Insufficient data.")


if __name__ == "__main__":
    interval = "1h"
    start_date = "01/01/2025"
    end_date = "05/02/2026"

    # Danh sách coin muốn chạy
    symbols = [
    "BTCUSDT","ETHUSDT","BNBUSDT","SOLUSDT","XRPUSDT","ADAUSDT","DOGEUSDT","AVAXUSDT","LINKUSDT",
    "DOTUSDT","LTCUSDT","BCHUSDT","TRXUSDT","ATOMUSDT","NEARUSDT","ETCUSDT","FILUSDT","APTUSDT","ARBUSDT",
    "OPUSDT","INJUSDT","SUIUSDT","SEIUSDT","TIAUSDT","XLMUSDT","UNIUSDT","AAVEUSDT","EGLDUSDT",
    "RUNEUSDT","IMXUSDT","GALAUSDT","DYDXUSDT","WIFUSDT","JUPUSDT","PYTHUSDT",
]

    initial_balance = 10000.0
    min_bars = 400

    summary_rows = []

    for symbol in symbols:
        print("\n" + "=" * 90)
        print(f"🪙 RUNNING: {symbol} | {interval} | {start_date} -> {end_date}")
        print("=" * 90)

        df_raw = fetch_binance_kline_df(symbol, interval, start_date, end_date)

        if df_raw is None or df_raw.empty or len(df_raw) <= min_bars:
            print(f"❌ Insufficient data for {symbol}. Bars={0 if df_raw is None else len(df_raw)}")
            summary_rows.append({
                "Symbol": symbol,
                "Status": "Insufficient data",
                "Final Equity": np.nan,
                "Total Return": np.nan,
                "CAGR": np.nan,
                "MaxDD": np.nan,
                "Sharpe": np.nan,
                "Trades": 0,
                "Winrate": np.nan,
                "ProfitFactor": np.nan,
            })
            continue

        print("✅ Data ready...")
        df_calc = calculate_indicators(df_raw)

        if df_calc.empty or len(df_calc) <= min_bars:
            print(f"❌ Insufficient after indicators for {symbol}. Bars={len(df_calc)}")
            summary_rows.append({
                "Symbol": symbol,
                "Status": "Insufficient after indicators",
                "Final Equity": np.nan,
                "Total Return": np.nan,
                "CAGR": np.nan,
                "MaxDD": np.nan,
                "Sharpe": np.nan,
                "Trades": 0,
                "Winrate": np.nan,
                "ProfitFactor": np.nan,
            })
            continue

        print("🚀 Running strategy...")
        bt = SMMA_RSI_KC_Breakout_Backtester(df_calc, initial_balance=initial_balance)
        bt.run()

        trades_df = pd.DataFrame(bt.trades)

        if not trades_df.empty:
            metrics = calculate_metrics(trades_df, bt.equity_curve, bt.initial_balance)
            print_report_and_plot(metrics, symbol)

            print("\n📊 Sample Trades:")
            print(trades_df.head(10).to_string())

            print("\n📈 Exit Reasons:")
            print(trades_df["Reason"].value_counts())

            summary_rows.append({
                "Symbol": symbol,
                "Status": "OK",
                "Final Equity": float(metrics["Final Equity"]),
                "Total Return": float(metrics["Total Return"]),
                "CAGR": float(metrics["CAGR (approx)"]),
                "MaxDD": float(metrics["Max Drawdown"]),
                "Sharpe": float(metrics["Sharpe"]),
                "Trades": int(metrics["Trades"]),
                "Winrate": float(metrics["Winrate"]),
                "ProfitFactor": float(metrics["Profit Factor"]),
            })
        else:
            print("⚠️ No trades.")
            # vẫn có equity_curve, nhưng metrics hiện tại dùng trades_df để tính PF/winrate
            # nên ta lưu cơ bản thôi
            summary_rows.append({
                "Symbol": symbol,
                "Status": "No trades",
                "Final Equity": bt.equity_curve[-1]["equity"] if bt.equity_curve else initial_balance,
                "Total Return": np.nan,
                "CAGR": np.nan,
                "MaxDD": np.nan,
                "Sharpe": np.nan,
                "Trades": 0,
                "Winrate": np.nan,
                "ProfitFactor": np.nan,
            })

    # ====== SUMMARY ======
    summary_df = pd.DataFrame(summary_rows)

    # sort theo Sharpe (nếu có), không có thì theo Final Equity
    if "Sharpe" in summary_df.columns:
        summary_df = summary_df.sort_values(by="Sharpe", ascending=False, na_position="last")
    else:
        summary_df = summary_df.sort_values(by="Final Equity", ascending=False, na_position="last")

    print("\n" + "#" * 90)
    print("📌 SUMMARY (sorted)")
    print("#" * 90)
    print(summary_df.to_string(index=False))
```

```python
import requests
import pandas as pd
import numpy as np
from datetime import datetime
import time
import matplotlib.pyplot as plt
import matplotlib.dates as mdates


# =========================
# 1) FETCH BINANCE DATA
# =========================
def str_to_millis(date_str: str) -> int:
    dt = datetime.strptime(date_str, "%d/%m/%Y")
    return int(dt.timestamp() * 1000)

def fetch_binance_kline_df(symbol: str, interval: str, start_str: str, end_str: str) -> pd.DataFrame:
    base_url = "https://api.binance.com/"
    api_url = "api/v3/klines"
    url = base_url + api_url

    start_time = str_to_millis(start_str)
    end_dt = datetime.strptime(end_str, "%d/%m/%Y").replace(hour=23, minute=59, second=59)
    end_time = int(end_dt.timestamp() * 1000)

    all_klines = []
    print(f"⏳ Downloading {symbol} {interval}...")

    while start_time < end_time:
        params = {
            "symbol": symbol.upper(),
            "interval": interval,
            "startTime": start_time,
            "endTime": end_time,
            "limit": 1000
        }
        try:
            r = requests.get(url, params=params, timeout=10)
            if r.status_code != 200:
                print("Error:", r.status_code, r.text)
                break

            data = r.json()
            if not data:
                break

            all_klines.extend(data)
            start_time = data[-1][0] + 1
            time.sleep(0.05)

        except Exception as e:
            print("Fetch error:", e)
            break

    if not all_klines:
        return pd.DataFrame()

    df = pd.DataFrame(all_klines, columns=[
        "timestamp", "Open", "High", "Low", "Close", "Volume",
        "close_time", "quote_asset_volume", "num_trades",
        "taker_buy_base_volume", "taker_buy_quote_volume", "ignore"
    ])
    df = df[["timestamp", "Open", "High", "Low", "Close", "Volume"]]
    df["timestamp"] = pd.to_datetime(df["timestamp"], unit="ms")
    df.set_index("timestamp", inplace=True)
    df = df.astype(float)
    return df


# =========================
# 2) INDICATORS
# =========================
def calculate_indicators(df: pd.DataFrame) -> pd.DataFrame:
    data = df.copy()

    # Median Price
    data["Median_Price"] = (data["High"] + data["Low"]) / 2.0

    # True Range
    prev_close = data["Close"].shift(1)
    tr = pd.concat([
        (data["High"] - data["Low"]),
        (data["High"] - prev_close).abs(),
        (data["Low"] - prev_close).abs(),
    ], axis=1).max(axis=1)
    data["TR"] = tr

    def calc_atr(tr_series: pd.Series, period: int) -> pd.Series:
        return tr_series.ewm(alpha=1/period, min_periods=period, adjust=False).mean()

    data["ATR_20"] = calc_atr(data["TR"], 20)
    data["ATR_30"] = calc_atr(data["TR"], 30)
    data["ATR_50"] = calc_atr(data["TR"], 50)

    def calc_rsi(series: pd.Series, period: int) -> pd.Series:
        delta = series.diff()
        gain = (delta.where(delta > 0, 0.0)).rolling(window=period).mean()
        loss = (-delta.where(delta < 0, 0.0)).rolling(window=period).mean()
        rs = gain / loss
        rsi = 100 - (100 / (1 + rs))
        return rsi

    data["RSI_50"] = calc_rsi(data["Median_Price"], 50)

    # Keltner Channel (20, dev=2.5)
    data["EMA_20"] = data["Close"].ewm(span=20, adjust=False).mean()
    data["ATR_20_KC"] = calc_atr(data["TR"], 20)
    data["KC_20_Upper"] = data["EMA_20"] + (2.5 * data["ATR_20_KC"])
    data["KC_20_Lower"] = data["EMA_20"] - (2.5 * data["ATR_20_KC"])

    # Keltner Channel (126, dev=2.5)
    data["EMA_126"] = data["Close"].ewm(span=126, adjust=False).mean()
    data["ATR_126"] = calc_atr(data["TR"], 126)
    data["KC_126_Upper"] = data["EMA_126"] + (2.5 * data["ATR_126"])
    data["KC_126_Lower"] = data["EMA_126"] - (2.5 * data["ATR_126"])

    # Highest/Lowest Close in 14 bars
    data["Highest_Close_14"] = data["Close"].rolling(window=14).max()
    data["Lowest_Close_14"] = data["Close"].rolling(window=14).min()

    data.dropna(inplace=True)
    return data


# =========================
# 3) BACKTESTER (Keltner + RSI + Stop + ATR + fee)
# =========================
class KeltnerRSIBacktester:
    def __init__(self, df: pd.DataFrame, initial_balance: float = 10000.0, fee_rate: float = 0.001):
        self.df = df
        self.initial_balance = float(initial_balance)
        self.balance = float(initial_balance)

        self.equity_curve = []
        self.trades = []

        self.position = None
        self.entry_price = 0.0
        self.sl = 0.0
        self.tp = 0.0
        self.entry_bar_idx = 0
        self.qty = 0.0

        self.pending_order = None
        self.trailing_sl = None
        self.trailing_activated = False

        # Fees
        self.fee_rate = float(fee_rate)
        self.total_fees = 0.0

        # Strategy Parameters
        self.risk_pct = 0.01
        self.sl_coef = 2.0
        self.tp_coef = 4.5
        self.price_entry_mult = 2.4
        self.exit_after_bars = 16

        # Trailing (fixed absolute points)
        self.trailing_stop = 90
        self.trailing_activation = 90

        self.pending_expiration = 9

    def calculate_fee(self, price: float, quantity: float) -> float:
        return float(price) * float(quantity) * self.fee_rate

    def run(self):
        idxs = self.df.index

        for i in range(len(self.df)):
            bar = self.df.iloc[i]
            t = idxs[i]

            # ---- Manage open position ----
            if self.position is not None:
                held = i - self.entry_bar_idx

                # Time exit
                if held >= self.exit_after_bars:
                    self.close_position(bar["Close"], t, "Time Exit")
                    self.append_equity(t)
                    continue

                if self.position == "LONG":
                    # trailing stop
                    if self.trailing_activated and self.trailing_sl is not None:
                        if bar["Low"] <= self.trailing_sl:
                            self.close_position(self.trailing_sl, t, "Trailing Stop")
                            self.append_equity(t)
                            continue

                    if bar["Low"] <= self.sl:
                        self.close_position(self.sl, t, "Stop Loss")
                        self.append_equity(t)
                        continue

                    if bar["High"] >= self.tp:
                        self.close_position(self.tp, t, "Take Profit")
                        self.append_equity(t)
                        continue

                    profit = bar["Close"] - self.entry_price
                    if profit >= self.trailing_activation:
                        self.trailing_activated = True
                        new_trailing_sl = bar["Close"] - self.trailing_stop
                        if self.trailing_sl is None or new_trailing_sl > self.trailing_sl:
                            self.trailing_sl = new_trailing_sl
                            self.sl = max(self.sl, self.trailing_sl)

                elif self.position == "SHORT":
                    # trailing stop
                    if self.trailing_activated and self.trailing_sl is not None:
                        if bar["High"] >= self.trailing_sl:
                            self.close_position(self.trailing_sl, t, "Trailing Stop")
                            self.append_equity(t)
                            continue

                    if bar["High"] >= self.sl:
                        self.close_position(self.sl, t, "Stop Loss")
                        self.append_equity(t)
                        continue

                    if bar["Low"] <= self.tp:
                        self.close_position(self.tp, t, "Take Profit")
                        self.append_equity(t)
                        continue

                    profit = self.entry_price - bar["Close"]
                    if profit >= self.trailing_activation:
                        self.trailing_activated = True
                        new_trailing_sl = bar["Close"] + self.trailing_stop
                        if self.trailing_sl is None or new_trailing_sl < self.trailing_sl:
                            self.trailing_sl = new_trailing_sl
                            self.sl = min(self.sl, self.trailing_sl)

            # ---- Manage pending order ----
            if self.position is None and self.pending_order is not None:
                if i >= self.pending_order["expire_idx"]:
                    self.pending_order = None
                else:
                    p = self.pending_order["price"]

                    if self.pending_order["type"] == "BUY_STOP" and bar["High"] >= p:
                        fill = max(p, bar["Open"])
                        self.open_position("LONG", fill, i, self.pending_order["sl_dist"], self.pending_order["tp_dist"])
                        self.pending_order = None

                    elif self.pending_order["type"] == "SELL_STOP" and bar["Low"] <= p:
                        fill = min(p, bar["Open"])
                        self.open_position("SHORT", fill, i, self.pending_order["sl_dist"], self.pending_order["tp_dist"])
                        self.pending_order = None

            # ---- Create signal (no position, no pending) ----
            if self.position is None and self.pending_order is None:
                if i < 3:
                    self.append_equity(t)
                    continue

                open_0 = bar["Open"]

                kc20_u_1 = self.df.iloc[i-1]["KC_20_Upper"]
                kc20_l_1 = self.df.iloc[i-1]["KC_20_Lower"]
                kc126_u_1 = self.df.iloc[i-1]["KC_126_Upper"]
                kc126_l_1 = self.df.iloc[i-1]["KC_126_Lower"]

                rsi_2 = self.df.iloc[i-2]["RSI_50"]
                rsi_1 = self.df.iloc[i-1]["RSI_50"]

                highest_3 = self.df.iloc[i-3]["Highest_Close_14"]
                lowest_3 = self.df.iloc[i-3]["Lowest_Close_14"]

                atr50_3 = self.df.iloc[i-3]["ATR_50"]
                atr20_0 = bar["ATR_20"]
                atr30_0 = bar["ATR_30"]

                long_signal = (open_0 > kc20_u_1) and (rsi_2 < rsi_1) and (open_0 > kc126_u_1)
                short_signal = (open_0 < kc20_l_1) and (rsi_2 > rsi_1) and (open_0 < kc126_l_1)

                if long_signal:
                    entry = highest_3 + self.price_entry_mult * atr50_3
                    sl_dist = self.sl_coef * atr20_0
                    tp_dist = self.tp_coef * atr30_0

                    self.pending_order = {
                        "type": "BUY_STOP",
                        "price": float(entry),
                        "sl_dist": float(sl_dist),
                        "tp_dist": float(tp_dist),
                        "expire_idx": i + self.pending_expiration
                    }

                elif short_signal:
                    entry = lowest_3 - self.price_entry_mult * atr50_3
                    sl_dist = self.sl_coef * atr20_0
                    tp_dist = self.tp_coef * atr30_0

                    self.pending_order = {
                        "type": "SELL_STOP",
                        "price": float(entry),
                        "sl_dist": float(sl_dist),
                        "tp_dist": float(tp_dist),
                        "expire_idx": i + self.pending_expiration
                    }

            self.append_equity(t)

    def append_equity(self, t):
        mtm = float(self.balance)
        close_px = float(self.df.loc[t, "Close"])

        if self.position == "LONG":
            mtm += (close_px - self.entry_price) * self.qty
        elif self.position == "SHORT":
            mtm += (self.entry_price - close_px) * self.qty

        self.equity_curve.append({"time": t, "equity": mtm})

    def open_position(self, pos_type: str, price: float, bar_idx: int, sl_dist: float, tp_dist: float):
        self.position = pos_type
        self.entry_price = float(price)
        self.entry_bar_idx = int(bar_idx)

        if pos_type == "LONG":
            self.sl = self.entry_price - float(sl_dist)
            self.tp = self.entry_price + float(tp_dist)
        else:
            self.sl = self.entry_price + float(sl_dist)
            self.tp = self.entry_price - float(tp_dist)

        self.trailing_sl = None
        self.trailing_activated = False

        # Risk-based qty
        risk_amount = self.balance * self.risk_pct
        sl_distance = abs(self.entry_price - self.sl)
        self.qty = (risk_amount / sl_distance) if sl_distance > 0 else 0.0

        # Fee at entry
        entry_fee = self.calculate_fee(self.entry_price, self.qty)
        self.balance -= entry_fee
        self.total_fees += entry_fee

    def close_position(self, price: float, t, reason: str):
        price = float(price)

        if self.position == "LONG":
            pnl = (price - self.entry_price) * self.qty
        else:
            pnl = (self.entry_price - price) * self.qty

        exit_fee = self.calculate_fee(price, self.qty)
        self.total_fees += exit_fee

        net_pnl = pnl - exit_fee
        self.balance += net_pnl

        self.trades.append({
            "Exit Time": t,
            "Type": self.position,
            "Entry": self.entry_price,
            "Exit": price,
            "PnL": pnl,
            "Fee": exit_fee,
            "Net PnL": net_pnl,
            "Reason": reason,
            "Balance": self.balance
        })

        self.position = None
        self.qty = 0.0
        self.trailing_sl = None
        self.trailing_activated = False


# =========================
# 4) METRICS (per coin)
# =========================
def calculate_advanced_metrics(trades_df: pd.DataFrame, equity_data, initial_balance: float, total_fees: float):
    eq_df = pd.DataFrame(equity_data).copy()
    eq_df.set_index("time", inplace=True)
    eq_df = eq_df[~eq_df.index.duplicated(keep="last")]

    final_equity = float(eq_df["equity"].iloc[-1])
    total_return = (final_equity - initial_balance) / initial_balance

    days = (eq_df.index[-1] - eq_df.index[0]).days
    years = days / 365.25 if days > 0 else 0.0
    cagr = ((final_equity / initial_balance) ** (1/years) - 1) if years > 0 else 0.0

    running_max = eq_df["equity"].cummax()
    dd = (eq_df["equity"] - running_max) / running_max
    max_dd = float(dd.min())

    eq_df["returns"] = eq_df["equity"].pct_change().fillna(0.0)
    mean_ret = eq_df["returns"].mean()
    std_ret = eq_df["returns"].std()

    bars_per_year = (len(eq_df) / years) if years > 0 else 0.0
    sharpe = float((mean_ret / std_ret) * np.sqrt(bars_per_year)) if std_ret != 0 and bars_per_year > 0 else 0.0

    total_trades = len(trades_df)
    win_trades = int((trades_df["Net PnL"] > 0).sum()) if total_trades > 0 else 0
    win_rate = (win_trades / total_trades) if total_trades > 0 else 0.0

    avg_pnl = float(trades_df["Net PnL"].mean()) if total_trades > 0 else 0.0
    gross_profit = float(trades_df.loc[trades_df["Net PnL"] > 0, "Net PnL"].sum()) if total_trades > 0 else 0.0
    gross_loss = float(abs(trades_df.loc[trades_df["Net PnL"] < 0, "Net PnL"].sum())) if total_trades > 0 else 0.0
    profit_factor = (gross_profit / gross_loss) if gross_loss > 0 else 0.0

    return {
        "Final Equity": final_equity,
        "Total Return": float(total_return),
        "CAGR (approx)": float(cagr),
        "Max Drawdown": float(max_dd),
        "Sharpe": float(sharpe),
        "Trades": int(total_trades),
        "Winrate": float(win_rate),
        "Avg PnL/Trade": float(avg_pnl),
        "Profit Factor": float(profit_factor),
        "Total Fees": float(total_fees),
        "equity_df": eq_df
    }


def run_backtest_for_symbol(symbol, interval, start_date, end_date, initial_balance, fee_rate, min_bars=300):
    print(f"\n{'='*80}")
    print(f"🔍 Symbol: {symbol} | {interval} | {start_date} -> {end_date}")
    print(f"{'='*80}")

    df_raw = fetch_binance_kline_df(symbol, interval, start_date, end_date)
    if df_raw is None or df_raw.empty or len(df_raw) < min_bars:
        print(f"❌ Insufficient data: bars={0 if df_raw is None else len(df_raw)}")
        return None

    df_calc = calculate_indicators(df_raw)
    if df_calc.empty or len(df_calc) < min_bars:
        print(f"❌ Insufficient after indicators: bars={len(df_calc)}")
        return None

    bt = KeltnerRSIBacktester(df_calc, initial_balance=initial_balance, fee_rate=fee_rate)
    bt.run()

    trades_df = pd.DataFrame(bt.trades)
    if trades_df.empty:
        print("⚠️ No trades.")
        return {
            "symbol": symbol,
            "backtester": bt,
            "metrics": {
                "Final Equity": bt.equity_curve[-1]["equity"] if bt.equity_curve else initial_balance,
                "Total Fees": bt.total_fees
            },
            "trades_df": trades_df
        }

    metrics = calculate_advanced_metrics(trades_df, bt.equity_curve, bt.initial_balance, bt.total_fees)

    print(f"✅ Done {symbol}: Final={metrics['Final Equity']:.2f} | Sharpe={metrics['Sharpe']:.3f} | "
          f"Trades={metrics['Trades']} | MaxDD={metrics['Max Drawdown']*100:.2f}%")
    return {
        "symbol": symbol,
        "backtester": bt,
        "metrics": metrics,
        "trades_df": trades_df
    }


# =========================
# 5) PORTFOLIO ENGINE (NO LOOKAHEAD)
# =========================
def build_aligned_equity_from_results(results):
    equity_series = {}
    for r in results:
        sym = r["symbol"]
        eq_df = pd.DataFrame(r["backtester"].equity_curve).copy()
        eq_df.set_index("time", inplace=True)
        eq_df = eq_df[~eq_df.index.duplicated(keep="last")]
        equity_series[sym] = eq_df["equity"]

    aligned_equity = pd.DataFrame(equity_series).sort_index()
    aligned_equity = aligned_equity.ffill().dropna(how="any")
    return aligned_equity


def calc_returns_from_equity(aligned_equity: pd.DataFrame) -> pd.DataFrame:
    return aligned_equity.pct_change().fillna(0.0)


def _normalize_and_clip_weights(w: pd.Series, w_max: float = None, eps: float = 1e-12) -> pd.Series:
    w = w.replace([np.inf, -np.inf], np.nan).fillna(0.0)
    s = float(w.sum())
    if s <= eps:
        n = len(w)
        return pd.Series(1.0 / n, index=w.index)

    w = w / s

    if w_max is not None:
        w = w.clip(upper=w_max)
        s2 = float(w.sum())
        if s2 <= eps:
            n = len(w)
            return pd.Series(1.0 / n, index=w.index)
        w = w / s2

    return w


def weights_equal(symbols):
    return pd.Series(1.0 / len(symbols), index=symbols)


def weights_inverse_vol(R_window: pd.DataFrame, eps: float = 1e-12):
    vol = R_window.std(ddof=0)
    inv = 1.0 / (vol + eps)
    return _normalize_and_clip_weights(inv)


def weights_momentum(R_window: pd.DataFrame):
    cum = (1.0 + R_window).prod(axis=0) - 1.0
    mom = cum.clip(lower=0.0)
    return _normalize_and_clip_weights(mom)


def weights_sharpe(R_window: pd.DataFrame, eps: float = 1e-12):
    mu = R_window.mean(axis=0)
    sd = R_window.std(axis=0, ddof=0)
    sharpe = (mu / (sd + eps)).clip(lower=0.0)
    return _normalize_and_clip_weights(sharpe)


def weights_risk_parity_cov(R_window: pd.DataFrame, max_iter=200, lr=0.05, eps=1e-12):
    cov = R_window.cov().values
    n = cov.shape[0]
    symbols = list(R_window.columns)

    w = np.ones(n) / n

    def port_vol(w_):
        return np.sqrt(np.maximum(w_ @ cov @ w_, eps))

    for _ in range(max_iter):
        pv = port_vol(w)
        mrc = (cov @ w) / (pv + eps)
        rc = w * mrc
        target = np.ones(n) * (pv / n)

        grad = (rc - target)
        w = w - lr * grad

        w = np.clip(w, 0.0, None)
        s = w.sum()
        if s <= eps:
            w = np.ones(n) / n
        else:
            w = w / s

    return pd.Series(w, index=symbols)


def get_weights_by_method(R_window: pd.DataFrame, method: str):
    if method == "equal":
        return weights_equal(R_window.columns)
    if method == "inverse_vol":
        return weights_inverse_vol(R_window)
    if method == "risk_parity":
        return weights_risk_parity_cov(R_window)
    if method == "momentum":
        return weights_momentum(R_window)
    if method == "sharpe":
        return weights_sharpe(R_window)
    raise ValueError(f"Unknown method: {method}")


def backtest_rebalanced_portfolio(
    R: pd.DataFrame,
    initial_capital: float = 10000.0,
    method: str = "equal",
    lookback: int = 60,
    rebalance_every: int = 168,   # H1 -> weekly
    w_max: float = 0.50
):
    symbols = list(R.columns)
    idx = R.index

    w_prev = weights_equal(symbols)
    w_hist = []
    port_ret = []

    for i in range(len(R)):
        # rebalance at fixed frequency, using ONLY past window (no lookahead)
        if i >= lookback and (i % rebalance_every == 0):
            window = R.iloc[i - lookback:i]
            w_new = get_weights_by_method(window, method)
            w_new = _normalize_and_clip_weights(w_new, w_max=w_max)
            w_prev = w_new

        r_i = float((R.iloc[i] * w_prev).sum())
        port_ret.append(r_i)
        w_hist.append(w_prev.values)

    port_ret = pd.Series(port_ret, index=idx, name="port_ret")
    port_equity = initial_capital * (1.0 + port_ret).cumprod()
    port_equity.iloc[0] = initial_capital

    port_df = pd.DataFrame({"Portfolio_Equity": port_equity, "port_ret": port_ret}, index=idx)
    w_df = pd.DataFrame(w_hist, index=idx, columns=symbols)
    return port_df, w_df


def portfolio_metrics(port_df: pd.DataFrame, initial_capital: float):
    eq = port_df["Portfolio_Equity"].dropna()
    final_equity = float(eq.iloc[-1])
    total_return = (final_equity - initial_capital) / initial_capital

    days = (eq.index[-1] - eq.index[0]).days
    years = days / 365.25 if days > 0 else 0.0
    cagr = ((final_equity / initial_capital) ** (1 / years) - 1) if years > 0 else 0.0

    running_max = eq.cummax()
    dd = (eq - running_max) / running_max
    max_dd = float(dd.min())

    r = port_df["port_ret"].fillna(0.0)
    mean_ret = r.mean()
    std_ret = r.std()

    bars_per_year = (len(r) / years) if years > 0 else 0.0
    sharpe = float((mean_ret / std_ret) * np.sqrt(bars_per_year)) if std_ret != 0 and bars_per_year > 0 else 0.0

    return {
        "final_equity": final_equity,
        "total_return": float(total_return),
        "cagr": float(cagr),
        "max_dd": float(max_dd),
        "sharpe": float(sharpe)
    }


def compare_methods(results, total_capital, rebalance_every=168, lookback=60, w_max=0.50):
    aligned_equity = build_aligned_equity_from_results(results)
    R = calc_returns_from_equity(aligned_equity)

    methods = {
        "Equal": "equal",
        "Inverse Vol": "inverse_vol",
        "Risk Parity (cov)": "risk_parity",
        "Momentum": "momentum",
        "Sharpe": "sharpe"
    }

    rows = []
    port_objects = {}

    for name, code in methods.items():
        print(f"\n⚙️ Testing method: {name}")
        port_df, w_df = backtest_rebalanced_portfolio(
            R,
            initial_capital=total_capital,
            method=code,
            lookback=lookback,
            rebalance_every=rebalance_every,
            w_max=w_max
        )
        met = portfolio_metrics(port_df, total_capital)
        rows.append({
            "Method": name,
            "Final Equity": met["final_equity"],
            "Total Return %": met["total_return"] * 100,
            "CAGR %": met["cagr"] * 100,
            "Max DD %": met["max_dd"] * 100,
            "Sharpe": met["sharpe"]
        })
        port_objects[name] = (port_df, w_df, aligned_equity)

    comp = pd.DataFrame(rows).sort_values(by="Sharpe", ascending=False)
    return comp, port_objects


def plot_portfolio(port_df, aligned_equity, weights_df, title):
    symbols = list(aligned_equity.columns)

    fig = plt.figure(figsize=(16, 12))
    gs = fig.add_gridspec(3, 1, height_ratios=[2, 1.5, 1])

    # Portfolio equity
    ax1 = fig.add_subplot(gs[0])
    ax1.plot(port_df.index, port_df["Portfolio_Equity"], linewidth=2.3)
    ax1.fill_between(port_df.index, port_df["Portfolio_Equity"], alpha=0.3)
    ax1.set_title(title, fontsize=16, fontweight="bold")
    ax1.set_ylabel("Equity ($)")
    ax1.grid(True, linestyle="--", alpha=0.3)
    ax1.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))

    # Individual equities
    ax2 = fig.add_subplot(gs[1])
    for sym in symbols:
        ax2.plot(aligned_equity.index, aligned_equity[sym], linewidth=1.2, label=sym)
    ax2.set_title("Individual Coin Equity Curves (net fee)", fontsize=13, fontweight="bold")
    ax2.set_ylabel("Equity ($)")
    ax2.legend(loc="upper left", fontsize=9, ncol=2)
    ax2.grid(True, linestyle="--", alpha=0.3)
    ax2.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))

    # Weights
    ax3 = fig.add_subplot(gs[2])
    ax3.stackplot(weights_df.index, *[weights_df[c] for c in symbols], labels=symbols, alpha=0.8)
    ax3.set_title("Portfolio Weights Over Time", fontsize=13, fontweight="bold")
    ax3.set_ylabel("Weight")
    ax3.set_ylim([0, 1])
    ax3.legend(loc="upper left", fontsize=9, ncol=len(symbols))
    ax3.grid(True, linestyle="--", alpha=0.3)
    ax3.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))

    plt.tight_layout()
    plt.show()


# =========================
# 6) PRINT METRICS LIKE IMAGE
# =========================
def compute_trade_stats_from_all_trades(results):
    """
    Gộp trades từ tất cả coin -> Trades / Winrate / Profit Factor
    Dùng Net PnL.
    """
    dfs = []
    for r in results:
        df = r.get("trades_df")
        if df is not None and not df.empty and "Net PnL" in df.columns:
            dfs.append(df)

    if not dfs:
        return {"Trades": 0, "Winrate": 0.0, "Profit Factor": 0.0}

    all_trades = pd.concat(dfs, ignore_index=True)

    total_trades = len(all_trades)
    win_trades = int((all_trades["Net PnL"] > 0).sum())
    winrate = win_trades / total_trades if total_trades > 0 else 0.0

    gross_profit = float(all_trades.loc[all_trades["Net PnL"] > 0, "Net PnL"].sum())
    gross_loss = float(abs(all_trades.loc[all_trades["Net PnL"] < 0, "Net PnL"].sum()))
    profit_factor = (gross_profit / gross_loss) if gross_loss > 0 else 0.0

    return {"Trades": int(total_trades), "Winrate": float(winrate), "Profit Factor": float(profit_factor)}


def print_metrics_like_image(metrics: dict, title: str = "METRICS"):
    """
    In metrics đúng format như ảnh bạn gửi.
    """
    print("\n" + "=" * 60)
    print(title)
    print("=" * 60)
    print(f"Final Equity: {metrics['Final Equity']:.6f}")
    print(f"Total Return: {metrics['Total Return']:.6f}")
    print(f"CAGR (approx): {metrics['CAGR (approx)']:.6f}")
    print(f"Max DrawDown: {metrics['Max Drawdown']:.6f}")
    print(f"Sharpe: {metrics['Sharpe']:.6f}")
    print(f"Trades: {int(metrics['Trades'])}")
    print(f"Winrate: {metrics['Winrate']:.6f}")
    print(f"Profit Factor: {metrics['Profit Factor']:.6f}")
    print("=" * 60)


# =========================
# 7) MAIN
# =========================
if __name__ == "__main__":
    # ===== Portfolio config =====
    # Bạn muốn 5 coin thì thêm vào list này. (hiện đang 4 coin như code bạn gửi)
    portfolio_symbols = ["ETHUSDT", "NEARUSDT", "ARBUSDT", "UNIUSDT"]

    interval = "1h"
    start_date = "01/01/2025"
    end_date = "22/02/2026"

    total_capital = 10000.0
    fee_rate = 0.001

    # Portfolio engine params
    lookback = 60
    rebalance_every = 168  # H1 -> weekly
    w_max = 0.50

    min_bars = 350

    print("=" * 90)
    print("🎯 PORTFOLIO BACKTEST (KeltnerRSI Strategy) - Rebalance Correctly (No Lookahead)")
    print("=" * 90)
    print(f"Total Capital: ${total_capital:.2f} | Fee rate: {fee_rate}")
    print(f"Symbols ({len(portfolio_symbols)}): {', '.join(portfolio_symbols)}")
    print(f"Interval: {interval} | {start_date} -> {end_date}")
    print(f"Lookback={lookback} | RebalanceEvery={rebalance_every} bars | w_max={w_max}")
    print("=" * 90)

    capital_per_coin = total_capital / len(portfolio_symbols)

    # ===== Run per-coin backtests =====
    results = []
    for sym in portfolio_symbols:
        res = run_backtest_for_symbol(
            sym, interval, start_date, end_date,
            initial_balance=capital_per_coin,
            fee_rate=fee_rate,
            min_bars=min_bars
        )
        if res is not None and res["backtester"].equity_curve:
            results.append(res)
        time.sleep(0.5)

    if len(results) < 2:
        print("\n⚠️ Not enough successful coin results to build portfolio (need >=2).")
        raise SystemExit

    # ===== Compare methods =====
    comp_df, port_objects = compare_methods(
        results,
        total_capital=total_capital,
        rebalance_every=rebalance_every,
        lookback=lookback,
        w_max=w_max
    )

    print("\n" + "=" * 90)
    print("📊 COMPARISON (sorted by Sharpe)")
    print("=" * 90)
    print(comp_df.to_string(index=False))

    # ===== Pick best and plot =====
    best_method = comp_df.iloc[0]["Method"]
    port_df, w_df, aligned_equity = port_objects[best_method]
    met = portfolio_metrics(port_df, total_capital)

    # ===== In metrics đúng kiểu ảnh =====
    trade_stats = compute_trade_stats_from_all_trades(results)
    metrics_like_image = {
        "Final Equity": met["final_equity"],
        "Total Return": met["total_return"],
        "CAGR (approx)": met["cagr"],
        "Max Drawdown": met["max_dd"],
        "Sharpe": met["sharpe"],
        "Trades": trade_stats["Trades"],
        "Winrate": trade_stats["Winrate"],
        "Profit Factor": trade_stats["Profit Factor"],
    }
    print_metrics_like_image(metrics_like_image, title=f"🏆 BEST PORTFOLIO METRICS - {best_method}")

    # ===== Plot =====
    plot_portfolio(
        port_df,
        aligned_equity,
        w_df,
        title=f"Portfolio Equity - {best_method} (Weekly Rebalance, Lookback={lookback})"
    )
```
