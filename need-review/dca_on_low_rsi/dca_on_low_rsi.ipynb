```python
import ccxt
import pandas as pd
import numpy as np
import talib as ta
import matplotlib.pyplot as plt

# Load historical data (example using Binance)
def fetch_data():
    exchange = ccxt.binance()
    bars = exchange.fetch_ohlcv('BTC/USDT', timeframe='1h', limit=500)
    df = pd.DataFrame(bars, columns=['timestamp', 'open', 'high', 'low', 'close', 'volume'])
    df['timestamp'] = pd.to_datetime(df['timestamp'], unit='ms')
    return df

# Calculate RSI
def calculate_rsi(df, period=14):
    df['rsi'] = ta.momentum.RSIIndicator(df['close'], window=period).rsi()
    return df

# Backtest Strategy
def backtest(df, initial_balance=1000):
    btc_holdings = initial_balance / df['close'].iloc[0]  # Buy and hold BTC
    balance = initial_balance  # Account balance
    short_position = 0  # Track short position
    short_price = 0  # Entry price of short

    for i in range(1, len(df)):
        rsi = df['rsi'].iloc[i]
        price = df['close'].iloc[i]

        # Enter short if RSI > 70
        if rsi > 70 and short_position == 0:
            short_position = balance / price  # Short equivalent BTC
            short_price = price
            print(f"Short Opened at ${price:.2f} (RSI: {rsi:.2f})")

        # Close short if RSI < 50
        elif rsi < 50 and short_position > 0:
            profit = short_position * (short_price - price)
            balance += profit
            print(f"Short Closed at ${price:.2f} (Profit: ${profit:.2f})")
            short_position = 0  # Reset short

    # Final portfolio value
    final_hold_value = btc_holdings * df['close'].iloc[-1]  # Buy & hold value
    print(f"Final Buy & Hold Value: ${final_hold_value:.2f}")
    print(f"Final Short Trading Balance: ${balance:.2f}")

# Run the strategy
df = fetch_data()
df = calculate_rsi(df)
backtest(df)
```
