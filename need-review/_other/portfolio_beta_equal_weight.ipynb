```python
import yfinance as yf
import pandas as pd
import numpy as np

def calculate_beta(ticker, benchmark, start_date, end_date):
    # Download historical data
    asset = yf.download(ticker, start=start_date, end=end_date)['Close']
    bench = yf.download(benchmark, start=start_date, end=end_date)['Close']

    # Calculate daily returns
    asset_returns = asset.pct_change().dropna()
    bench_returns = bench.pct_change().dropna()

    # Align data
    data = pd.concat([asset_returns, bench_returns], axis=1).dropna()
    data.columns = [ticker, benchmark]

    # Calculate covariance and variance
    covariance = np.cov(data[ticker], data[benchmark])[0, 1]
    variance = np.var(data[benchmark])

    # Calculate beta
    beta = covariance / variance
    return beta

# Set parameters
start_date = '2018-01-01'
end_date = '2023-09-12'
commodities = [
    "CL=F",   # Crude Oil WTI
    "BZ=F",   # Crude Oil Brent
    "NG=F",   # Natural Gas
    "GC=F",   # Gold
    "SI=F",   # Silver
    "PL=F",   # Platinum
    "HG=F",   # Copper
    "PA=F",   # Palladium
    "ZC=F",   # Corn
    "ZW=F",   # Wheat
    "ZS=F",   # Soybeans
    "KC=F",   # Coffee
    "CC=F",   # Cocoa
    "CT=F",   # Cotton
    "SB=F",   # Sugar
    "LE=F",   # Live Cattle
    "HE=F"    # Lean Hogs
]
currencies = [
    "EURUSD=X",  # Euro / US Dollar
    "JPY=X",     # US Dollar / Japanese Yen
    "GBPUSD=X",  # British Pound / US Dollar
    "AUDUSD=X",  # Australian Dollar / US Dollar
    "NZDUSD=X",  # New Zealand Dollar / US Dollar
    "USDCAD=X",  # US Dollar / Canadian Dollar
    "USDCHF=X",  # US Dollar / Swiss Franc
    "USDCNY=X",  # US Dollar / Chinese Yuan
    "USDHKD=X",  # US Dollar / Hong Kong Dollar
    "USDSGD=X"   # US Dollar / Singapore Dollar
]
ticker = commodities + currencies
benchmark = 'SPY'  # S&P 500 ETF

# Calculate betas
betas = {}
for t in ticker:
    beta = calculate_beta(t, benchmark, start_date, end_date)
    betas[t] = beta
    print(f"Beta for {t}: {beta}")
    betas_df = pd.DataFrame(list(betas.items()), columns=['Ticker', 'Beta'])
    betas_df.sort_values(by='Beta', ascending=False, inplace=True)
    # betas_df.to_csv('betas.csv', index=False)
betas_df
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import timedelta

# Define parameters
tickers = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'TSLA', 'NVDA', 'JPM', 'WMT', 'PG', 'KO',
           'XOM', 'CVX', 'GE', 'BA', 'MMM', 'CAT', 'DIS', 'NKE', 'MCD', 'VZ']  # Example stocks
benchmark = 'SPY'  # S&P 500 ETF
start_date = '2018-01-01'  # Extended start for 1-year lookback
end_date = '2025-09-12'
rolling_window = 252  # 1-year lookback (trading days)
rebalance_period = 21  # Approx 1 month (trading days)
risk_free_rate = 0.02  # Annual risk-free rate (2%)
trading_days = 252  # Trading days per year

# Download close prices
tickers.append(benchmark)
data = yf.download(tickers, start=start_date, end=end_date)['Close']
data = data.dropna()

# Calculate daily returns
daily_returns = data.pct_change().dropna()

# Function to calculate rolling beta
def calculate_rolling_beta(stock_returns, bench_returns, window):
    betas = []
    for i in range(window, len(stock_returns)):
        cov_matrix = np.cov(stock_returns[i-window:i], bench_returns[i-window:i])
        covariance = cov_matrix[0, 1]
        variance = np.var(bench_returns[i-window:i])
        beta = covariance / variance if variance != 0 else np.nan
        betas.append(beta)
    return pd.Series(betas, index=stock_returns.index[window:])

# Calculate rolling betas for all stocks
beta_dfs = {}
for ticker in tickers[:-1]:  # Exclude SPY
    beta_dfs[ticker] = calculate_rolling_beta(daily_returns[ticker], daily_returns[benchmark], rolling_window)

# Combine betas into a DataFrame
beta_df = pd.DataFrame(beta_dfs, index=daily_returns.index[rolling_window:])

# Initialize portfolio
portfolio_returns = []
portfolio_dates = []
selected_stocks = []
current_date = beta_df.index[0]
end_date = pd.to_datetime(end_date)

while current_date <= beta_df.index[-1]:
    # Get betas for the current rebalance date
    current_betas = beta_df.loc[current_date]

    # Select top 5 lowest and top 5 highest beta stocks
    lowest_beta = current_betas.nsmallest(5).index
    highest_beta = current_betas.nlargest(5).index
    portfolio_tickers = list(lowest_beta) + list(highest_beta)

    # Equal weights (10% each for 10 stocks)
    weights = [0.1] * 10

    # Calculate portfolio returns until next rebalance or end
    next_date = min(current_date + timedelta(days=rebalance_period), beta_df.index[-1])
    period_returns = daily_returns.loc[current_date:next_date, portfolio_tickers]
    period_portfolio_returns = (period_returns * weights).sum(axis=1)

    portfolio_returns.extend(period_portfolio_returns)
    portfolio_dates.extend(period_portfolio_returns.index)
    selected_stocks.append((current_date, portfolio_tickers))

    # Move to next rebalance date
    current_date = beta_df.index[beta_df.index > current_date][0] if current_date < beta_df.index[-1] else beta_df.index[-1]

# Create portfolio returns series
portfolio_returns = pd.Series(portfolio_returns, index=portfolio_dates)

# Calculate cumulative returns
cumulative_returns = (1 + portfolio_returns).cumprod()

# Calculate annualized return
total_return = cumulative_returns[-1] - 1
days = (cumulative_returns.index[-1] - cumulative_returns.index[0]).days
years = days / 365.25
annualized_return = (1 + total_return) ** (1 / years) - 1

# Calculate annualized volatility
portfolio_volatility = portfolio_returns.std() * np.sqrt(trading_days)

# Calculate Sharpe ratio (annualized)
mean_daily_return = portfolio_returns.mean()
std_daily_return = portfolio_returns.std()
sharpe_ratio = (mean_daily_return * trading_days - risk_free_rate) / (std_daily_return * np.sqrt(trading_days))

# Calculate maximum drawdown
rolling_max = cumulative_returns.cummax()
drawdowns = (cumulative_returns - rolling_max) / rolling_max
max_drawdown = drawdowns.min()

# Print results
print(f"Portfolio Annualized Return: {annualized_return:.2%}")
print(f"Portfolio Annualized Volatility: {portfolio_volatility:.2%}")
print(f"Portfolio Sharpe Ratio (annualized): {sharpe_ratio:.2f}")
print(f"Portfolio Maximum Drawdown: {max_drawdown:.2%}")
print(f"Portfolio Cumulative Return: {(cumulative_returns[-1] - 1):.2%}")
print("\nSample of selected stocks (Date, [Low Beta, High Beta]):")
for date, stocks in selected_stocks[:3]:  # Show first 3 rebalances
    print(f"{date.date()}: {stocks}")

# Plot cumulative returns
plt.figure(figsize=(10, 6))
plt.plot(cumulative_returns.index, cumulative_returns, label='Portfolio Cumulative Return')
plt.title('High/Low Beta Portfolio Cumulative Returns')
plt.xlabel('Date')
plt.ylabel('Cumulative Return')
plt.grid(True)
plt.legend()
plt.show()

# Plot drawdowns
plt.figure(figsize=(10, 6))
plt.plot(drawdowns.index, drawdowns, label='Portfolio Drawdown', color='red')
plt.title('Portfolio Drawdown')
plt.xlabel('Date')
plt.ylabel('Drawdown')
plt.grid(True)
plt.legend()
plt.show()
```
