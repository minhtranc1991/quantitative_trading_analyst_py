```python
import requests
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime
import numpy as np

```

```python
# Dogecoin API from CoinGecko
url = "https://api.coingecko.com/api/v3/coins/dogecoin/market_chart"
params = {
    "vs_currency": "usd",
    "days": "1",  # Last 30 days of data
    "interval": "daily"
}

response = requests.get(url, params=params)
data = response.json()
coin = "dogecoins"
# Print data or save to CSV
# for price in data['prices']:
#     print(f"Timestamp: {price[0]}, Price: {price[1]}")
```

```python
timestamps = [datetime.utcfromtimestamp(price[0] / 1000) for price in data['prices']]  # Convert timestamps to datetime
prices = [price[1] for price in data['prices']]  # Price data
market_caps = [market_cap[1] for market_cap in data['market_caps']]  # Market cap data
volumes = [volume[1] for volume in data['total_volumes']]  # Volume data

# Calculate percentage change between consecutive prices
percentage_changes = [(prices[i] - prices[i - 1]) / prices[i - 1] * 100 for i in range(1, len(prices))]

volume_market_cap_ratio = [volumes[i] / market_caps[i] for i in range(1, len(volumes))]
```

```python
x_data = volume_market_cap_ratio[:-1]  # V/M ratio of the previous day
y_data = percentage_changes[1:]  # Percentage change of the current day
```

```python
# Plotting the data
plt.figure(figsize=(10, 6))
plt.scatter(x_data, y_data, color='tab:blue', alpha=0.7)

# Adding labels and title
plt.title(f"{coin} Volume-to-Market-Cap Ratio (Previous Day) vs. Price Change (Current Day)", fontsize=14)
plt.xlabel("Volume-to-Market-Cap Ratio (Previous Day)", fontsize=12)
plt.ylabel("Percentage Price Change (Current Day)", fontsize=12)
plt.grid(True)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
percentage_changes_abs = [abs((prices[i] - prices[i - 1]) / prices[i - 1]) * 100 for i in range(1, len(prices))]
y_abs_data = percentage_changes_abs[1:]
correlation = np.corrcoef(x_data, y_abs_data)[0, 1]

print(f"Pearson Correlation between Volume-to-Market-Cap Ratio (Previous Day) and Price Change (Current Day): {correlation:.4f}")
```
