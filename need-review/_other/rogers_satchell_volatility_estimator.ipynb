```python
import numpy as np
import pandas as pd

def rogers_satchell_volatility(df):
  """
	Rogers–Satchell volatility estimator.
	df: DataFrame với cột 'Open', 'High', 'Low', 'Close'
	"""
  term1 = np.log(df['High'] / df['Open']) * np.log(df['High'] / df['Close'])
  term2 = np.log(df['Low'] / df['Open']) * np.log(df['Low'] / df['Close'])
  rs_var = (term1 + term2).mean()
  return np.sqrt(rs_var)

# Ví dụ
data = {
	'Open':  [100, 102, 101],
	'High':  [105, 104, 103],
	'Low':   [99, 100, 99],
	'Close': [104, 101, 102]
	}
df = pd.DataFrame(data)
print("Rogers–Satchell Volatility:", rogers_satchell_volatility(df))
```
